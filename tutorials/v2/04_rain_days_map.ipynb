# 04 · Rain days map: the v2 record on an interactive page

Builds `rain_days_map/openmesh_rain_days.html` and shows it here. Two tabs:

- **Record overview**: the numbers of the v2 record, its three sampling periods, paths reporting per day, and the metadata tables.
- **Rain days map**: two rain days, 2024-03-23 and 2025-10-30. Every link coloured by its peak attenuation, the 60 and 24 GHz
  links with a rain rate (the chain of `03_showcase`), ASOS gauges and MRMS hourly radar. Click a link to see its
  attenuation or rain rate against the gauges within the radius.

Data, all downloaded on the first run: the v2 record (Zenodo 23283445), the Weather Underground PWS of the **v1** record
(Zenodo 17508286; v2 has no PWS, so only 2024-03-23 has them) and Natural Earth land. The page is not stored in the
repository because it embeds the v1 PWS data (CC BY-NC 4.0); this notebook writes it.
